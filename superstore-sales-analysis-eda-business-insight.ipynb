# 🛒 Superstore Sales Analysis: E-commerce Performance & Insights

## 📌 1. Objetivo do Projeto
Este projeto tem como objetivo analisar o desempenho de vendas, lucratividade e hábitos de compra de uma grande loja de departamentos (Superstore). 
Através desta análise exploratória, buscaremos responder às seguintes perguntas de negócio:
1. Quais regiões geram mais vendas e lucros?
2. Quais categorias de produtos apresentam margem de lucro baixa ou prejuízo?
3. Qual o impacto dos descontos nas vendas e no lucro total?
4. Quais são os modos de envio mais utilizados pelos clientes?

```python
# Importação das bibliotecas essenciais para manipulação e visualização de dados
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração visual dos gráficos
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)

# Carregando o arquivo do dataset disponível no ambiente do Kaggle
import os
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        if filename.endswith('.csv'):
            data_path = os.path.join(dirname, filename)

# Lendo o arquivo CSV
df = pd.read_csv(data_path, encoding='latin1')

# Exibindo as primeiras 5 linhas do dataset
df.head()
```

## 🧹 2. Limpeza e Inspeção dos Dados
Vamos verificar as dimensões do dataset, tipos de colunas, valores nulos e possíveis registros duplicados.

```python
# Informações gerais das colunas
print("--- Informações dos Dados ---")
df.info()

print("\n--- Valores Nulos ---")
print(df.isnull().sum())

print("\n--- Linhas Duplicadas ---")
print(df.duplicated().sum())

# Removendo duplicadas se houver
if df.duplicated().sum() > 0:
    df = df.drop_duplicates()
    print("\nLinhas duplicadas removidas com sucesso!")
```

## 📊 3. Análise Exploratória de Dados (EDA)

### 3.1. Vendas e Lucro por Categoria e Subcategoria de Produtos

```python
# Agrupando Vendas e Lucro por Categoria
category_summary = df.groupby('Category')[['Sales', 'Profit']].sum().reset_index()

# Plotando gráfico de barras lado a lado
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(data=category_summary, x='Category', y='Sales', ax=ax[0], palette='viridis')
ax[0].set_title('Vendas Totais por Categoria ($)')
ax[0].set_ylabel('Vendas ($)')

sns.barplot(data=category_summary, x='Category', y='Profit', ax=ax[1], palette='crest')
ax[1].set_title('Lucro Total por Categoria ($)')
ax[1].set_ylabel('Lucro ($)')

plt.tight_layout()
plt.show()
```

### 3.2. Impacto dos Descontos no Lucro
Vamos verificar se a concessão de descontos maiores melhora ou prejudica a lucratividade da empresa.

```python
# Gráfico de linha mostrando a relação entre Desconto e Lucro
plt.figure(figsize=(10, 5))
sns.lineplot(data=df, x='Discount', y='Profit', color='red', marker='o')
plt.title('Relação entre Taxa de Desconto e Lucro Médio')
plt.xlabel('Taxa de Desconto (%)')
plt.ylabel('Lucro Médio ($)')
plt.axhline(0, color='black', linestyle='--', linewidth=0.8) # Linha do zero (ponto de equilíbrio)
plt.show()
```

### 3.3. Desempenho de Vendas e Lucro por Região

```python
# Agrupando por Região
region_summary = df.groupby('Region')[['Sales', 'Profit']].sum().reset_index()

plt.figure(figsize=(10, 5))
sns.barplot(data=region_summary, x='Region', y='Profit', palette='Blues_d')
plt.title('Lucro Total por Região Geográfica')
plt.xlabel('Região')
plt.ylabel('Lucro Total ($)')
plt.show()
```

## 💡 4. Principais Insights e Recomendações de Negócio

1. **Atenção aos Descontos Elevados:** Descontos acima de 20% impactam severamente a margem da loja, gerando lucro médio negativo (prejuízo). Recomenda-se limitar a concessão de descontos a no máximo 15%.
2. **Categoria de Tecnologia:** É o segmento mais lucrativo da loja. Deve-se investir em campanhas de marketing focadas nesses produtos.
3. **Móveis (Furniture):** Embora apresente alto volume de vendas, a categoria de móveis possui baixa margem de lucro devido a custos operacionais e altos descontos. Reavaliação de fornecedores e logística é recomendada.
